# **DP Pipeline**

```python
from pyspark import pipelines as dp 
from pyspark.sql.functions import *
```

### **Streaming Table**

```python
# Expectation
expectations = {
    'rule1':'product_id is not null',
    'rule2':'product_name is not null'
}
```

```python
@dp.table()
@dp.expect_all_or_drop(expectations)
def DimProducts_stage():
    return spark.readStream.table('ecommerce.silver.products')
```

### **Materialized View**

```python
@dp.materialized_view()

def DimProducts_materialized_view():
    return spark.read.table('Live.DimProducts_stage')
```

### **DimProducts**

```python
dp.create_streaming_table('DimProducts')
```

```python
dp.create_auto_cdc_flow(
  target = "DimProducts",
  source = "Live.DimProducts_materialized_view",
  keys = ["product_id"],
  sequence_by = col("product_id"),
  stored_as_scd_type = 2
)
```
