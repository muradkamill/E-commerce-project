```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

### **Making dynamic for all**

```python
dbutils.widgets.text('file_name','')
```

```python
file_name=dbutils.widgets.get('file_name')
```

### **Data reading with structured streaming (Autolader)**

```python
inf_df=spark.readStream\
    .format('cloudFiles').option('cloudFiles.format','parquet')\
    .option('cloudFiles.schemaLocation',f'abfss://bronze@datalakecommerce.dfs.core.windows.net/checkpoint_{file_name}')\
    .load(f'abfss://source@datalakecommerce.dfs.core.windows.net/{file_name}')
```

```python
inf_df.writeStream.format('parquet')\
    .outputMode('append')\
    .option('checkpointLocation',f'abfss://bronze@datalakecommerce.dfs.core.windows.net/checkpoint_{file_name}')\
    .option('path',f'abfss://bronze@datalakecommerce.dfs.core.windows.net/{file_name}')\
    .trigger(once=True)\
    .start()
```
